```python
df_formality_examples = df.copy()

df_formality_examples["formality"] = pd.to_numeric(
    df_formality_examples["formality"],
    errors="coerce"
)

df_formality_examples = df_formality_examples.dropna(
    subset=["formality", "text"]
)

# Highest and lowest formality examples
highest_formality = (
    df_formality_examples
    .sort_values("formality", ascending=False)
    .iloc[0]
)

lowest_formality = (
    df_formality_examples
    .sort_values("formality", ascending=True)
    .iloc[26]         
)

# Print function
def print_formality_example(row, title):
    print("\n" + "=" * 90)
    print(title)
    print("=" * 90)
    print("Speech ID:", row.get("Speech_id", "NA"))
    print("Party:", row.get("party", "NA"))
    print("Person:", row.get("person", "NA"))
    print("Date:", row.get("date", "NA"))
    print("Formality:", row["formality"])
    print("\nTEXT:")
    print(row["text"])

print_formality_example(
    highest_formality,
    "HIGHEST FORMALITY EXAMPLE IN CORPUS"
)

print_formality_example(
    lowest_formality,
    "LOWEST FORMALITY / HIGHEST INFORMALITY EXAMPLE IN CORPUS"
)
```
